## 6. Point-in-time interpretation

An observation period is not the same as public availability. A monthly or quarterly observation must not be treated as known at the end of its observation period unless publication evidence supports that assumption.

Where historical release timestamps or vintages are missing:

- preserve the original observation period;
- do not invent publication timestamps;
- apply only an explicitly documented conservative availability policy;
- retain the assigned PIT classification;
- do not claim strict PIT backtesting eligibility for unvalidated series.

The Phase 3 report records a conservative availability policy of 10:00 on the next London business day for the BoE database series. This policy must be interpreted within the scope and assumptions documented in that report; it is not a substitute for missing historical vintage evidence.
