## 1. Notebook purpose

Document the Phase 1 source investigation retrospectively: source categories, provider candidates, evaluation criteria, the rationale for the initial research universe, licensing and point-in-time (PIT) constraints, and the acquisition direction subsequently reflected in Phase 2.

This is a documentary research record. It does not claim that every candidate provider was tested, that a complete provider comparison was performed, or that every proposed dataset was approved or acquired.

## 2. Project context

- Research architecture: multi-asset and provider-independent.
- Pilot / target instrument: National Grid (`NG.L`).
- Initial comparable utilities: SSE (`SSE.L`) and Severn Trent (`SVT.L`).
- Initial broad market benchmark: FTSE 100 (`^FTSE`).
- Intended market-data frequency: daily; lower-frequency macroeconomic series retain their native frequency.
- Base currency: GBP.
- Intended use: research, backtesting and paper trading.
- Live-money trading and broker integration are outside the approved initial scope.

**Evidence status:** The configuration is supported by the Master Project Specification and subsequent Phase 2 acquisition records. The complete original Phase 1 deliberation sequence is not available in the evidence reviewed.

## 3. Research questions

1. Which sources can provide historical equity market data for the pilot and comparable assets?
2. Which official sources provide UK monetary, yield-curve and macroeconomic observations?
3. What historical depth, frequency, metadata and revision history are available?
4. Can historical information availability be established sufficiently for PIT research?
5. What are the API, reliability, cost and licensing constraints?
6. Which sources may be suitable for exploratory research, strict PIT research or eventual commercial use?

These questions describe the investigation framework. They are not evidence that every question was answered through a completed provider test.

## 5. Evaluation framework

The source investigation was intended to consider:

- Historical coverage and frequency.
- Data accuracy and completeness.
- API availability, limits and reliability.
- Observation, publication, retrieval and revision timestamps.
- Historical vintages and PIT availability.
- Licensing, redistribution and commercial-use rights.
- Cost and operational complexity.
- Reproducibility, provenance and auditability.

**Important distinction:** a source may be usable for exploratory research while remaining unsuitable for strict PIT backtesting or commercial distribution. Free access or technical availability does not establish a right to redistribute data or commercialise derived products.

## 6. National Grid research rationale

NG.L was selected as the pilot because the project requires a concrete UK-listed utility target while retaining a reusable multi-asset architecture.

The initial explanatory research universe was intended to investigate:

- UK interest rates and gilt yields, given their potential relationship with financing costs and equity valuation;
- inflation and GDP indicators, subject to publication-time and vintage controls;
- comparable utilities (SSE and Severn Trent);
- broad UK equity-market context (FTSE 100);
- energy, FX, volatility, credit, regulatory and company-specific information as hypotheses for later investigation.

These are **economic hypotheses**, not established predictive relationships. In particular, energy commodity prices must not be assumed to map directly to National Grid equity returns. Any variable group must demonstrate incremental out-of-sample value before it is retained.

## 7. Point-in-time and licensing principles

- Record the observation period separately from publication, retrieval and revision timestamps.
- Never use a macroeconomic observation before it was publicly available.
- Respect revisions and historical vintages where available.
- If release timestamps or vintages are unavailable, do not invent them; apply only an explicitly documented conservative availability policy.
- Keep PIT eligibility separate from licensing eligibility.
- Do not infer commercial rights from free access or API availability.
- The FRED/ALFRED training restriction remains in force pending written authorisation expressly resolving it.

PIT validity does not imply licensing permission, and licensing permission does not establish PIT validity.
